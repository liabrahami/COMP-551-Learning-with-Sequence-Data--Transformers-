```python
from datasets import load_dataset
import random
```

## Task 1: Data split

```python
dataset = load_dataset("fancyzhx/ag_news")

print(dataset) # see 120k train instances and 7.6k test instances

whole_train_dataset = dataset["train"]
test_dataset = dataset["test"]
```

```python
SEED = 2026

generator = random.seed(SEED) ############# verify if this makes sense

split = whole_train_dataset.train_test_split(
    test_size = 0.2,
    seed = generator,
    stratify_by_column = "label" # get same proportion of labels/classes in subsets
)

train_dataset = split["train"]
val_dataset = split["test"]

print(train_dataset)
print(val_dataset)
```
